```python
import pandas as pd
df = pd.read_csv("heart.csv")
df.head()
```

```python
import numpy,pandas,sklearn
import matplotlib, seaborn
print("ML stack done")
```

```python
df.info()
df.isnull().sum()
```

```python
df["target"].value_counts()
```

```python
df.describe()
```

```python
x = df.drop("target",axis=1)
y = df["target"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000)
lr.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = lr.predict(x_test)

print("Accuracy: ", accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train,y_train)
```

```python
y_pred_dt = dt.predict(x_test)

print("Decision Tree Accuracy:", accuracy_score(y_test,y_pred_dt))
print(classification_report(y_test,y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=42)
rf.fit(x_train,y_train)
```

```python
y_pred_rf = rf.predict(x_test)

print("Random Forest Classifier Accuracy:", accuracy_score(y_test,y_pred_rf))
print(classification_report(y_test,y_pred_rf))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.neural_network import MLPClassifier

nn = MLPClassifier(hidden_layer_sizes=(32,16),max_iter=1000,random_state=42)

nn.fit(x_train_scaled,y_train)
```

```python
y_pred_nn = nn.predict(x_test_scaled)

print("Neural Network Accuracy:", accuracy_score(y_test,y_pred_nn))
print(classification_report(y_test,y_pred_nn))
```

```python
import pandas as pd

results = pd.DataFrame({
    "Model" : ["Logistic Regression", "Decision Tree", "Random Forest","Neural Network"],
    "Accuracy" : [
        accuracy_score(y_test,y_pred),
        accuracy_score(y_test,y_pred_dt),
        accuracy_score(y_test,y_pred_rf),
        accuracy_score(y_test,y_pred_nn)
        
    ] 
})

results
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.title("Logistic Regression - Confusion matrix")
plt.show
```

```python
from sklearn.metrics import roc_curve,auc

y_prob = lr.predict_proba(x_test)[:,1]
fpr,tpr,_=roc_curve(y_test,y_prob)
roc_auc =auc(fpr ,tpr)

import matplotlib.pyplot as plt

plt.plot (fpr,tpr,label = "AUC = %0.2f" % roc_auc)
plt.plot([0,1],[0,1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Logistic Regression")
plt.legend()
plt.show()
```

## Feature Importance Explanation

Feature importance represents how strongly each medical feature contributes to the prediction of heart disease. In this dataset, important contributing features include age, chest pain type, maximum heart rate achieved, oldpeak, slope, number of major vessels, and thalassemia condition. These features play a major role in determining whether a patient is likely to have heart disease.

In tree based models such as Decision Tree and Random Forest, feature importance is calculated based on how well a feature helps split the data and reduce impurity. In Logistic Regression, feature influence is determined using model coefficients, which show how strongly each feature affects the final prediction probability.

Understanding feature importance improves the transparency of the model and helps medical professionals trust the system by knowing which health indicators impact the prediction the most. This also adds strong real world relevance to the project.
